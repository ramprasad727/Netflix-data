```python
df['duration']=df['duration'].fillna(df['duration'].mode()[0])
```

```python
df['director']=df['director'].fillna(df['director'].mode()[0])
```

```python
df['country']=df['country'].fillna(df['country'].mode()[0])
```

```python
df['cast']=df['cast'].fillna(df['cast'].mode()[0])
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.drop(df[['date_added']],axis=1)
```

```python
df['date_added']=pd.to_datetime(df['date_added'],format='%d %m %Y',errors='coerce')
```

```python
df.head()
```

```python
df['Year']=df['date_added'].dt.year
df['Month']=df['date_added'].dt.month
```

```python
df.head()
```

```python
df['type'].value_counts()
```

```python
plt.figure(figsize=(7,5))
ax=sns.countplot(x='type',data=df)
ax.bar_label(ax.containers[0])
plt.title('Count of Cinema Type')
plt.show()
```

```python
a=df['release_year'].value_counts().sort_index()
a
```

```python
plt.figure(figsize=(7,5))
plt.plot(a,data=df)
plt.title('Content Released Each Year')
plt.xlabel('Year')
plt.ylabel('Count')
plt.show()
```
